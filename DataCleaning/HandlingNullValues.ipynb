```python
df = df.drop(columns=['Dimensions','Plot Area'])
```

```python
df.isnull().sum()
```

```python
df['Car Parking'] = df['Car Parking'].fillna('No Parking')
df['Balcony'] = df['Balcony'].fillna('No Balcony')
df['Furnishing'] = df['Furnishing'].fillna('Unfurnished')
df['facing'] = df['facing'].fillna('Unknown')
df['Ownership'] = df['Ownership'].fillna('Unknown')
df['Society'] = df['Society'].fillna('Standalone/Independent')
df['overlooking'] = df['overlooking'].fillna('Unknown')
```

```python
df['total_area'] = df['Super Area'].fillna(df['Carpet Area'])
```

```python
df = df.drop(columns=['Super Area','Carpet Area'])
```

```python
df['Price (in rupees)'] = df['Price (in rupees)'].fillna(df['Price (in rupees)'].median())
```

```python
df['Bathroom'] = pd.to_numeric(df['Bathroom'],errors='coerce')
```

```python
df['Bathroom'] = df['Bathroom'].fillna(df['Bathroom'].median())
```

```python
df['Floor'].info()
```

```python
df['Status'] = df['Status'].fillna(df['Status'].mode()[0])
df['Transaction'] = df['Transaction'].fillna(df['Transaction'].mode()[0])
df['Floor'] = df['Floor'].fillna(df['Floor'].mode()[0])
df['Description'] = df['Description'].fillna('No Description')
```

```python
df['total_area'] = pd.to_numeric(df['total_area'],errors='coerce')
```

```python
df['total_area'] = df['total_area'].fillna(df['total_area'].median())
```
